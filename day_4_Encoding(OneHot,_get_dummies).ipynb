```python
df=pd.read_csv("/content/covid_toy.csv")
```

```python
a= pd.get_dummies(df, columns=['gender','cough','city'])
a
```

```python
a['has_covid'] = a['has_covid'].map({
    'Yes': 1,
    'No': 0
})
```

```python
a=a.astype(int)
a
```

```python
b=pd.get_dummies(df, columns =['gender','cough','city'],drop_first=True)
b
```

```python
b=b.astype(int)
b
```
