```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(drop='first',
             sparse_output=False, dtype=np.int32)
```

```python
ohe_arr= ohe.fit_transform(df[['sex','smoker','region']])
```

```python
ohe_df=pd.DataFrame(ohe_arr,columns=ohe.get_feature_names_out(),
                    index=df.index)
ohe_df
```

```python
df=pd.concat([df.drop(['sex','smoker','region'], axis=1), ohe_df], axis=1)
```

```python
df.head()
```

```python
df.shape
```

```python
x= df.drop(columns=['charges'])
y=df['charges']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# model selection and training
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr= LinearRegression()
```

```python
lr.fit(x_train, y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
# Performance Evalution Metrics
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```
