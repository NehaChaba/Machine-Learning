```python
df=pd.read_csv("/content/Churn_Modelling.csv")
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
lb=LabelEncoder()
```

```python
df['Surname']=lb.fit_transform(df['Surname'])
df['Geography']=lb.fit_transform(df['Geography'])
df['Gender']=lb.fit_transform(df['Gender'])
```

```python
df.head(2)
```

```python
x=df.drop(columns=['Exited'])
y=df['Exited']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
# Ordinal encoding
```

```python
df=pd.read_csv("/content/Churn_Modelling.csv")
```

```python
df.head(2)
```

```python
df.drop(columns ='Surname')
```

```python
df=df[['Geography','Gender']]
```

```python
df.head(2)
```

```python
df['Geography'].value_counts()
```

```python
df['Gender'].value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe=OrdinalEncoder(categories=[
    ['France','Germany','Spain'],
    ['Male','Female']
])
```

```python
df_arr=oe.fit_transform(df)
df_arr
```

```python
df_new=pd.DataFrame(df_arr,columns=df.columns,dtype=int)
df_new
```
