```python
# Normal Distribution ----> Data mean = 0, Standard Deviation=1
# Central tendency(mean,mode,median) ---> left side and right side equal data points
```

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("/content/insurance.csv")
df.head(2)
```

```python
df['sex']=df['sex'].map({'female':1,
                         'male':0})
```

```python
df.isnull().sum()
```

```python
df['smoker']=df['smoker'].map({'yes':1,
                         'no':0})
```

```python
df['region'].value_counts()
```

```python
df['region']=df['region'].map({'southeast':0,
                         'southwest':1,
                          'northeast':2,
                          'northwest':3})
```

```python
df.sample(4)
```

```python
x=df.drop(columns=['charges'])
y=df['charges']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
np.round(x_train.describe(),2)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc=StandardScaler()
```

```python
x_train_sc=sc.fit_transform(x_train)
```

```python
x_train_new=pd.DataFrame(x_train_sc,columns=x_train.columns)
```

```python
np.round(x_train_new.describe(),2)
```

```python
# Normalisation (min_value=0,max_value=1)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
mn=MinMaxScaler()
```

```python
x_train_mn = mn.fit_transform(x_train)
```

```python
x_train_new = pd.DataFrame(x_train_mn,columns=x_train.columns)
```

```python
np.round(x_train_new.describe(),2)
```
